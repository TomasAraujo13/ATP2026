```python
# a) Lista formada pelos elementos que não são comuns às duas listas:

lista1 = [1, 2, 3, 4, 5]
lista2 = [4, 5, 6, 7, 8]  
ncomuns = [elem for elem in lista1 if elem not in lista2] + [elem for elem in lista2 if elem not in lista1]
print(f"{ncomuns}")

# Resultado esperado: [1,2,3,7,8]
```

```python
# b) Lista formada pelas palavras do texto compostas por mais de 3 letras:

texto = """Vivia há já não poucos anos algures num concelho do Ribatejo 
    um pequeno lavrador e negociante de gado chamado Manuel Peres Vigário"""
lista = [palavra for palavra in texto.split() if len(palavra) > 3]
print(f"{lista}")

# Resultado esperado: ['Vivia', 'poucos', 'anos', 'algures', 'concelho', ...]
```

```python
# c)  Lista formada por pares do tipo (índice, valor) com os valores da lista dada:

lista = ['anaconda', 'burro', 'cavalo', 'macaco']
listaRes = [(i, lista[i - 1]) for i in range(1, len(lista) + 1)]
print(f"{listaRes}")

# Resultado esperado: [(1,'anaconda'), (2,'burro'), (3,'cavalo'), (4,'macaco')]
```

```python
# a) Especifique uma função que dada uma string e uma substring não vazia, calcula  o número de vezes em que a substring aparece na string, sem que haja sobreposição de substrings:

def strCount(s, subs):
    return s.count(subs)

print(strCount("catcowcat", "cat")) # --> 2
print(strCount("catcowcat", "cow")) # --> 1
print (strCount("catcowcat", "dog")) # --> 0
```

```python
# b)  Especifique uma função que recebe uma lista de números inteiros positivos e devolve o menor produto que for possível calcular multiplicando os 3 menores inteiros da lista:

def produtoM3(lista):
    lista_ordenada = sorted(lista)
    return lista_ordenada[0] * lista_ordenada [1] * lista_ordenada [2]

print(produtoM3([12,3,7,10,12,8,9]))
# Resultado esperado: 168 = 3 * 7 * 8
```

```python
# c) Especifique uma função que dado um número inteiro positivo, repetidamente adiciona os seus dígitos até obter apenas um dígito que é retornado como resultado:

# Input: 38
# Output: 2
# Explicação: 3 + 8 = 11, 1 + 1 = 2.

# Input: 777
# Output: 3
# Explicação: 7 + 7 + 7 = 21, 2 + 1 = 3.
def reduxInt(n):
    while n >= 10:
        n = sum(int(digito) for digito in str(n))
    return n
print(reduxInt(67)) # 6 + 7 = 13, 1 + 3 = 4
```

```python
# d) Especifique uma função que recebe duas strings, `string1` e `string2`, e devolve o índice da primeira ocorrência de `string2` em `string1`, caso não ocorra nenhuma vez a função deverá retornar `-1`:

#Invocação: indexOf("Hoje está um belo dia de sol!", "belo")
# Resultado: 13

# Invocação: indexOf("Hoje está um belo dia de sol!", "chuva")
# Resultado: -1

def myIndexOf(s1, s2):
    res = -1  # se não encontrar nada, devolvo -1
    i = 0

    while i <= len(s1) - len(s2) and res == -1:
        if s1[i:i + len(s2)] == s2:
            res = i
        i = i + 1
    return res 

print(myIndexOf("Hoje está um belo dia de sol!", "belo"))
print(myIndexOf("Hoje está um belo dia de sol!", "chuva"))
```

```python
MyFaceBook = [
    {'id': 'p1', 'conteudo':  'A tarefa de avaliação é talvez a mais ingrata das tarefas que um professor tem de realizar...', 'autor': 'jcr', 'dataCriacao': '2023-07-20', 'comentarios': [
            {'comentario': 'Completamente de acordo...','autor': 'prh'}, {'comentario': 'Mas há quem goste...','autor': 'jj'} ]},
    {'id': 'p2', 'conteudo': 'O Benfica é o maior de Portugal', 'autor': 'Tomás Araújo', 'dataCriacao': '2026-11-01', 'comentarios': []}
]
```

```python
# a) `quantosPost`, que indica quantos posts estão registados:

def quantosPost(redeSocial):
    return len(redeSocial)
print(quantosPost(MyFaceBook))
```

```python
# b) `postsAutor`, que devolve a lista de posts de um determinado autor:

def postsAutor(redeSocial, autor):
    return [p for p in redeSocial if p['autor'] == autor] 
print(postsAutor(MyFaceBook,'jcr'))
```

```python
# c) `autores`, que devolve a lista de autores de posts ordenada alfabeticamente:

def autores(redeSocial):
    lista_autores = []

    for p in redeSocial:
        autor = p['autor']
        if autor not in lista_autores:
            lista_autores.append(autor)

    return  sorted(lista_autores)

print(autores(MyFaceBook))
```

```python
# d) `insPost`, que acrescenta um novo post à rede social a partir dos parâmetros recebidos e devolve a nova rede social. 

def insPost(redeSocial, conteudo, autor, dataCriacao, comentarios):
    novo_id = "p" + str(len(redeSocial) + 1)

    novo_post = {

        'id': novo_id, 
        'conteudo': conteudo, 
        'autor': autor, 
        'dataCriacao': dataCriacao, 
        'comentarios': comentarios
     }

    redeSocial.append(novo_post)

    return redeSocial

print(insPost(MyFaceBook,"O meu gato é fofo","Sofia Aguiar", "2026-11-01","A minha gata também é fofa" ))
```

```python
# e) `remPost`, que remove um post da rede, correspondente ao `id` recebido.

def remPost(redeSocial, id):
    return [p for p in redeSocial if p['id'] != id]

print(remPost(MyFaceBook, 'p1'))
```

```python
# f) `postsPorAutor`, que devolve uma distribuição de posts por autor.

def postsPorAutor(redeSocial):
    contagem = {}
    
    for p in redeSocial:
        autor = p['autor']
       
        if autor in contagem:
            contagem[autor] += 1
        else:
            contagem[autor] = 1
            
    return contagem
print(postsPorAutor(MyFaceBook))
```
